# Carga de librerias y dataset de entrenamiento Para uso Exploratorio

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
%matplotlib inline

df =pd.read_csv("data/fraudTrain.csv")
```

```python
# Info del dataset
df.info()
```

```python
# Utilizamos .head() para ver los primeros 5 registros del dataset
df.head()
```

```python
# Utilizamos el .Describe()  para ver completamente los tipos de datos y media , mediana y std
df.describe()
```

```python
# Verificacion de datos faltanes en dataset
df.isna().sum()
```

```python
# Verificacion de datos nulos
print(df.isnull().sum())
```

```python
# validando promedio media y desviacion estandar de la colimna objetivo is_fraud
promedio_fraude=df["is_fraud"].mean()
mediana_fraude=df["is_fraud"].median()
std_fraude=df["is_fraud"].std()
print(f"media:{promedio_fraude}\nmediana:{mediana_fraude}\nstd{std_fraude}")
```

```python
# 1. Convertision de datetime especificando el formato exacto
# El formato ''%Y-%m-%d %H:%M:%S'
df['trans_date_trans_time'] = pd.to_datetime(df['trans_date_trans_time'], format='%Y-%m-%d %H:%M:%S')

# 2. Ahora sí extraer la hora como número entero
hora_dia= df['hora_dia'] = df['trans_date_trans_time'].dt.hour

# Verificar que sea numérico
print(df['hora_dia'].dtype) # Debería decir int32 o int64
print(df['hora_dia'].head())
```

```python
print(df['category'].isnull().sum())
print(df['category'].unique())
print(df.columns.tolist())
```

```python
# --- GRÁFICA 1: is_fraud vs category ---

plt.style.use('default') 
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 12

# Preparamos los datos: Contamos las ocurrencias por categoría y tipo de fraude
# Esto crea una tabla pivote para facilitar el gráfico de barras agrupadas
tabla_categoria = df.groupby(['category', 'is_fraud']).size().unstack(fill_value=0)

# Graficamos
ax1 = tabla_categoria.plot(kind='bar', color=['green', 'red'], width=0.8)

plt.title('Distribución de Fraude por Categoría de Transacción')
plt.xlabel('Categoría')
plt.ylabel('Número de Transacciones')
plt.legend(title='Es Fraude', labels=['No (0)', 'Sí (1)'])
plt.xticks(rotation=45, ha='right') # Rotar etiquetas para que se lean bien
plt.tight_layout()
plt.show()

```

```python
# --- GRÁFICA 2: is_fraud vs trans_date_trans_time ---

# Preparamos los datos: Contamos fraudes y no-fraudes por hora
tabla_hora = df.groupby([hora_dia, 'is_fraud']).size().unstack(fill_value=0)

# Graficamos
ax2 = tabla_hora.plot(kind='line', marker='o', color=['green', 'red'], linewidth=2)

plt.title('Tendencia de Fraude según la Hora del Día')
plt.xlabel('Hora del Día (0-23)')
plt.ylabel('Número de Transacciones')
plt.legend(title='Es Fraude', labels=['No (0)', 'Sí (1)'])
plt.xticks(range(0, 24)) # Asegurar que se vean todas las horas
plt.grid(True, linestyle='--', alpha=0.5) # Cuadrícula suave para leer mejor
plt.tight_layout()
plt.show()
```
